# Entangled Pair with Spectator Qubits

`q0` and `q1` form a Bell pair via `H` + `CX`. The `CCX(q1, q2, q3)` never fires because `q2` is always `|0⟩`, so `q2` and `q3` stay in `|0⟩` while `q0`–`q1` collapse together: `|0000⟩` or `|0011⟩`.

```python
from qiskit import QuantumCircuit
from qiskit.primitives import StatevectorSampler

qc = QuantumCircuit(4)
qc.h(0)
qc.cx(0, 1)
qc.ccx(1,2,3)
qc.measure_all()

sampler = StatevectorSampler()
result = sampler.run([qc], shots=1024).result()
print("State: Bell pair (q0,q1) ⊗ |00⟩ (q2,q3 spectators)")
print(result[0].data.meas.get_counts())
```

# Bell States

### |Φ+⟩ = (|00⟩ + |11⟩) / √2

No `X` gates before `H(q0)` + `CX(q0,q1)` — starting from `|00⟩`.

```python
from qiskit import QuantumCircuit
from qiskit.primitives import StatevectorSampler

qc = QuantumCircuit(2)
qc.h(0)
qc.cx(0, 1)
qc.measure_all()

print("State: |Φ+⟩ = (|00⟩ + |11⟩) / √2")
print("Circuit Diagram:")
print(qc.draw(output='text'))
sampler = StatevectorSampler()
result = sampler.run([qc], shots=1000).result()
print(result[0].data.meas.get_counts())
```

### |Φ-⟩ = (|00⟩ - |11⟩) / √2

`X(q0)` before `H(q0)` + `CX(q0,q1)` flips the relative phase — starting from `|10⟩`.

```python
from qiskit import QuantumCircuit
from qiskit.primitives import StatevectorSampler

qc = QuantumCircuit(2)
qc.x(0)
qc.h(0)
qc.cx(0, 1)
qc.measure_all()

print("State: |Φ-⟩ = (|00⟩ - |11⟩) / √2")
print("Circuit Diagram:")
print(qc.draw(output='text'))
sampler = StatevectorSampler()
result = sampler.run([qc], shots=1000).result()
print(result[0].data.meas.get_counts())
```

### |Ψ+⟩ = (|01⟩ + |10⟩) / √2

`X(q1)` before `H(q0)` + `CX(q0,q1)` — starting from `|01⟩`.

```python
from qiskit import QuantumCircuit
from qiskit.primitives import StatevectorSampler

qc = QuantumCircuit(2)
qc.x(1)
qc.h(0)
qc.cx(0, 1)
qc.measure_all()

print("State: |Ψ+⟩ = (|01⟩ + |10⟩) / √2")
print("Circuit Diagram:")
print(qc.draw(output='text'))
sampler = StatevectorSampler()
result = sampler.run([qc], shots=1000).result()
print(result[0].data.meas.get_counts())
```
